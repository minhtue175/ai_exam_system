## 2. Mô phỏng Đường cong Quên lãng Ebbinghaus & Hệ thống Hộp Leitner
Mô phỏng xác suất ghi nhớ lại kiến thức của học viên sau các khoảng thời gian ôn tập lặp lại ngắt quãng (Spaced Repetition) từ Ngăn 1 đến Ngăn 5.

```python
days = np.linspace(0, 30, 300)

# Ebbinghaus Forgetting Curve without review: R = e^(-t/S)
r_no_review = np.exp(-days / 3.0) * 100

# Spaced Repetition reviews at Day 1, Day 3, Day 7, Day 14
r_spaced = np.zeros_like(days)
for i, t in enumerate(days):
    if t < 1:
        r_spaced[i] = np.exp(-t / 3.0) * 100
    elif t < 3:
        r_spaced[i] = np.exp(-(t - 1) / 7.0) * 100
    elif t < 7:
        r_spaced[i] = np.exp(-(t - 3) / 16.0) * 100
    elif t < 14:
        r_spaced[i] = np.exp(-(t - 7) / 35.0) * 100
    else:
        r_spaced[i] = np.exp(-(t - 14) / 75.0) * 100

plt.figure(figsize=(10, 5))
plt.plot(days, r_no_review, label="Không ôn tập (Ebbinghaus decay)", color="#dc3545", linestyle="--", linewidth=2)
plt.plot(days, r_spaced, label="Leitner Spaced Repetition (Fluesy Exam)", color="#6a35ff", linewidth=2.5)
plt.title("Hiệu quả ghi nhớ dài hạn với Dual-Mode Leitner Spaced Repetition", fontsize=14, pad=12)
plt.xlabel("Thời gian (Ngày)", fontsize=11)
plt.ylabel("Tỷ lệ ghi nhớ (% Retained Memory)", fontsize=11)
plt.ylim(0, 105)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()
plt.show()
```
